```python
import numpy as np
import pandas as pd
df = pd.DataFrame({'salary': [65, 75, 95, 110, 85]})

# Define conditions
conditions = [
    (df['salary'] < 70),
    (df['salary'] < 90),
    (df['salary'] >= 90)
]

# Define corresponding labels
labels = ['brokie', 'poor', 'average']

# Create new column
df['salary_group'] = np.select(conditions, labels, default='')

print(df)
```

```python
df = pd.read_csv("ds_salaries.csv")
ax = df.groupby("job_title").mean("salary_in_usd")['salary_in_usd'].sort_values().plot.bar(figsize=(20, 10))
ax.set_xlabel('Job Title')
ax.set_ylabel('Salary in USD')
ax.set_title('Average Salary by Job Title')
plt.show()
```

```python
ax = df.sort_values(by='salary_in_usd').plot.scatter(x="job_title", y="salary_in_usd", figsize=(20, 10))
ax.set_xticks(range(len(df['job_title'].unique())))
ax.set_xticklabels(df.sort_values(by='salary_in_usd')['job_title'].unique(), rotation=90)
ax.set_xlabel('Job Title')
ax.set_ylabel('Salary in USD')
ax.set_title('Salary Prediction')
plt.show()
```

```python
lowexp=df[df['experience_level']== 'EN']
midexp=df[df['experience_level']== 'MI']
highexp=df[df['experience_level'] == 'SE']
expertexp=df[df['experience_level'] == 'EX']

smallcomp=df[df['company_size']== 'S']
midcomp=df[df['company_size']== 'M']
largecomp=df[df['company_size'] == 'L']
```

```python
fig, axes = plt.subplots(4, 2, figsize=(10, 8))

#------entry----
axes[0,0].hist(lowexp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='Entry Level')
axes[0,0].set_xlabel('Entry Level')

#-----mids
axes[1,0].hist(midexp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='Mid Level')
axes[1,0].set_xlabel('Mid Level')


#-----senior---

axes[2,0].hist(highexp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='Senior Level')
axes[2,0].set_xlabel('Senior Level')

#-----expert
axes[3, 0].hist(expertexp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='Expert Level')
axes[3, 0].set_xlabel('Expert Level')

#------small comp
axes[0, 1].hist(smallcomp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='small company')
axes[0, 1].set_xlabel('Small company')

#------medium comp
axes[1, 1].hist(midcomp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='Medium company')
axes[1, 1].set_xlabel('Medium company')

#------large comp
axes[2, 1].hist(largecomp["salary_in_usd"], bins=20, color='red', alpha=0.5, label='large company')
axes[2, 1].set_xlabel('Large company')



fig.suptitle('Distributions by Pay', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert categorical data
exp = df['experience_level'].astype('category').cat.codes
comp = df['company_size'].astype('category').cat.codes
sal = df['salary_in_usd']

# Bar sizes
dx = np.ones_like(exp) * 0.5
dy = np.ones_like(comp) * 0.5
dz = sal

# Plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(exp, comp, np.zeros_like(sal), dx, dy, dz)

ax.set_xlabel("Experience Level")
ax.set_ylabel("Company Size")
ax.set_zlabel("Salary (USD)")

plt.tight_layout()
plt.show()
```
